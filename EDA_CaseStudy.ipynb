```python
#Now we will display for Piechart
trace = go.Pie(
    labels=['Dead', 'Alive'],
    values=[549,342],
    marker=dict(colors=['pink', 'hotpink'])
)
layout = go.Layout(title='Survived Distribution')
fig = go.Figure(data=trace, layout=layout)
fig.show()
```

```python
#Box plot and Scatter Plot
trace =  go.Box(x=data['Pclass'],
                y=data['Age'],
                marker={'color': 'hotpink'})
layout = go.Layout(title="Age V/S Pclass Distribution")
fig = go.Figure(data=trace, layout= layout)
fig.show()
```

```python
#Scatter Plot representation --> Age V/S Fare
trace = go.Scatter(x=data['Age'],y=data['Fare'],
                   mode='markers',
                   marker={'color':'pink'})
layout = go.Layout(
    title="Age V/S Fare Distribution")
fig = go.Figure(data=trace,layout=layout)
fig.show()
```

```python
#Understanding the missing values and filtering the missng values 
data.isnull().sum()
```

```python
print(data['Age'].mean())
```

```python
#We can fill the age column missing with mean
data['Age'] = data['Age'].fillna(data['Age'].mean(),inplace=True)
data['Age'].isnull().sum()
```

```python
data['Embarked'] = data['Embarked'].fillna('S', inplace=True)
```

```python
data.drop(columns=['Cabin'],inplace=True)
```
